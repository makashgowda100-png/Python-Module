## Operators¶
    Arithmetic Operators : +,-,*,/,power
    Logical Operators : AND,OR,NOT
    Membership Operators : IN, NOT IN
    Conditional Operators : >,<,==,>=,<=
    Assignment Operators : +=,-=,*=,...

```python
# Arithmetic Operators
# Addition
a = 10
b = 20
add1 = a+b
print(add1)
```

```python
# Subtraction
sub1 = b-a
print(sub1)
```

```python
# Multiplication
Mul1=(a*b)
print(Mul1)
```

```python
# Division
div1=(a/b)
print(div1)
```

```python
# Integer Division
int_div = a//b
print(int_div)
```

```python
# Modulus(Gives reminder)
Rem_div = a%b
print(Rem_div)
```

```python
# power
pow_div = a**b
print(pow_div)
```
